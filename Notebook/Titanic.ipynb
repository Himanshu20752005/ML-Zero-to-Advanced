```python
import sklearn
print(sklearn.__version__)
from sklearn.model_selection import train_test_split
```

```python
# train test split
x = df.drop(["Survived"], axis=1)
y = df["Survived"]
```

```python
x_train , x_test , y_train  , y_test = train_test_split(x,y, test_size = 0.2, random_state=42) 
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
x_train.isnull().sum()
```

```python
"""140 missing out of 712 is significant, but we don't want to throw away those passengers.

We'll probably use the median age to fill the missing values.

Why median instead of mean?

Because age can have some unusually high values, and the median is less affected by those outliers."""
```

```python
# age_median = df["Age"].median()
# print(age_median)

#median tho train pe hi loge na 

age_median = x_train["Age"].median()
print(age_median)
```

```python
x_train["Age"] = x_train["Age"].fillna(age_median)
x_train.isnull().sum()
```

```python
embark = x_train["Embarked"].mode()[0]
print(embark)
""" it return a series of mode ("Since there can be multiple modes, like A,A,B,B,C,C,
    the mode will be A,B,C") so we take the first value of that series so the [0] is 
    used to get the first value (index)"""
```

```python
print(x_train["Embarked"])
x_train["Embarked"] = x_train["Embarked"].fillna(embark)

print(x_train.isnull().sum())
```

```python
x_train = x_train.drop("Cabin",axis = 1)
x_test = x_test.drop("Cabin",axis = 1)
```
