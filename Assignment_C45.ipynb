```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold

data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain', 'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Strong'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}
df = pd.DataFrame(data)
print(df)
```

```python
def entropy(df, t):
    counts = df[t].value_counts(normalize=True)
    ent = 0
    for c in counts:
        ent -= c * np.log2(c)
    return ent

print("Entropy of Play:", entropy(df, "Play"))
```

```python
def information_gain(df, feature, target):
    total_entropy = entropy(df, target)
    weighted_entropy = 0
    for value in df[feature].unique():
        subset = df[df[feature] == value]
        weight = len(subset) / len(df)
        weighted_entropy += weight * entropy(subset, target)
    return total_entropy - weighted_entropy
```

```python
def split_info(df,feature):
    s_info = 0
    for value in df[feature].unique():
        subset = df[df[feature] == value ]
        s_info -= (len(subset)/len(df))*np.log2(len(subset)/len(df))
    return s_info        
```

```python
def gain_ratio(df,feature, target):
    ig = information_gain(df, feature, target)
    si = split_info(df,feature)
    if si == 0:
        return 0
    return ig/si    
 
```

```python
for col in df.columns: 
    if col != "Play": 
        print(col, ":", gain_ratio(df, col, "Play"))
```
